### Respuesta al inciso c.iii: Consideraciones para abrir el archivo en Python

Al abrir un archivo CSV con Python (por ejemplo, mediante `pandas.read_csv`), hay que tener en cuenta:

1. **Codificación de caracteres (Encoding):** Como se pudo comprobar al ejecutar el código, el archivo no utiliza `UTF-8` estándar sino `latin-1` (ISO-8859-1) debido a caracteres especiales y acentos. Si no se especifica `encoding='latin-1'`, Python arroja un error de tipo `UnicodeDecodeError`.
2. **Delimitadores y comas internas:** Columnas como `cast`, `title` o `description` contienen comas dentro del mismo texto. Se debe asegurar que el parser interprete correctamente las comillas dobles (`"`) para evitar que divida una misma columna en varias.
3. **Manejo de datos faltantes (NaN):** El dataset contiene celdas vacías o nulas (en directores, actores, países), por lo que Pandas las cargará automáticamente como valores `NaN` (`Not a Number`).
4. **Tipos de datos asignados:** Pandas infiere los tipos de datos al leer el archivo. Columnas como `release_year` las tomará como entero, mientras que fechas como `date_added` se cargarán como texto (`object`), requiriendo una conversión posterior si se quieren realizar análisis temporales.
