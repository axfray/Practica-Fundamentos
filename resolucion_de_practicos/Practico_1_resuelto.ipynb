## 2. Suma de dos numeros

```python
def sumar(numero_a, numero_b):
    return numero_a + numero_b

sumar(7, 5)
```

## 3. Area de un circulo

```python
def area_circulo(radio):
    if radio < 0:
        raise ValueError('El radio no puede ser negativo')
    return pi * radio ** 2

area_circulo(3)
```

## 5. Numero par o impar

```python
def es_par(numero):
    return numero % 2 == 0

for numero in [8, 11]:
    resultado = 'par' if es_par(numero) else 'impar'
    print(f'{numero} es {resultado}')
```

## 6. Promedio de notas

```python
notas_estudiantes = {'Ana': 8, 'Bruno': 7, 'Carla': 10, 'Diego': 9}
promedio = sum(notas_estudiantes.values()) / len(notas_estudiantes)
print(f'Promedio: {promedio:.2f}')
```

## 7. Primeros 20 numeros de Fibonacci

```python
def fibonacci(cantidad):
    numeros = []
    anterior, siguiente = 0, 1
    for _ in range(cantidad):
        numeros.append(anterior)
        anterior, siguiente = siguiente, anterior + siguiente
    return numeros

fibonacci(20)
```

## 8. Palindromo

```python
def es_palindromo(texto):
    texto_normalizado = texto.lower().replace(' ', '')
    return texto_normalizado == texto_normalizado[::-1]

print(es_palindromo('neuquen'))
print(es_palindromo('python'))
```

## 9. Clase Rectangulo

```python
class Rectangulo:
    def __init__(self, ancho, alto):
        if ancho < 0 or alto < 0:
            raise ValueError('Las dimensiones no pueden ser negativas')
        self.ancho = ancho
        self.alto = alto

    def area(self):
        return self.ancho * self.alto

    def perimetro(self):
        return 2 * (self.ancho + self.alto)

rectangulo = Rectangulo(4, 6)
print(f'Area: {rectangulo.area()}')
print(f'Perimetro: {rectangulo.perimetro()}')
```

## 10. Clase Cuadrado y herencia

```python
class Cuadrado(Rectangulo):
    def __init__(self, lado):
        super().__init__(lado, lado)

cuadrado = Cuadrado(5)
print(f'Area del cuadrado: {cuadrado.area()}')
print(f'Perimetro del cuadrado: {cuadrado.perimetro()}')
```
