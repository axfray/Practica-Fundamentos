# Practico 2: Acceso a datos

Se descargan, leen e inspeccionan tres conjuntos de datos. El orden de trabajo es: descargar, leer, revisar tipos y faltantes, transformar cuando hace falta y recien despues analizar.

```python
from pathlib import Path
import re
import pandas as pd
from IPython.display import display
from wget import download

DATA_DIR = Path('datos')
DATA_DIR.mkdir(exist_ok=True)

def descargar_si_falta(url, nombre):
    destino = DATA_DIR / nombre
    if not destino.exists():
        print(f'Descargando {nombre}...')
        download(url, out=str(destino))
    else:
        print(f'Ya existe {nombre}')
    return destino

URLS = {
    'obesity_txt': 'https://ignaciorlando.github.io/datasets/data-science/ObesityDataSet_raw_and_data_sinthetic.txt',
    'obesity_csv': 'https://ignaciorlando.github.io/datasets/data-science/ObesityDataSet_raw_and_data_sinthetic.csv',
    'netflix_txt': 'https://ignaciorlando.github.io/datasets/data-science/netflix_titles.txt',
    'netflix_csv': 'https://ignaciorlando.github.io/datasets/data-science/netflix_titles.csv',
    'catalogo_xlsx': 'https://ignaciorlando.github.io/datasets/data-science/catalogo_streaming_argentina_2024.xlsx',
}

archivos = {
    clave: descargar_si_falta(url, nombre)
    for clave, (url, nombre) in {
        'obesity_txt': (URLS['obesity_txt'], 'obesity_description.txt'),
        'obesity_csv': (URLS['obesity_csv'], 'obesity.csv'),
        'netflix_txt': (URLS['netflix_txt'], 'netflix_description.txt'),
        'netflix_csv': (URLS['netflix_csv'], 'netflix_titles.csv'),
        'catalogo_xlsx': (URLS['catalogo_xlsx'], 'catalogo_streaming_argentina_2024.xlsx'),
    }.items()
}

print('Archivos disponibles:', sorted(path.name for path in DATA_DIR.iterdir()))
```

## 1. ObesityDataSet_raw_and_data_sinthetic

```python
descripcion_obesidad = archivos['obesity_txt'].read_text(encoding='utf-8')
print(descripcion_obesidad)
```

```python
obesity = pd.read_csv(archivos['obesity_csv'])
print('Dimensiones:', obesity.shape)
display(obesity.head())
display(obesity.head(5))
display(obesity.describe(include='all').T)
```

### Tipos y significado de las columnas

Las variables numericas no son todas continuas. Algunas representan categorias o cantidades discretas codificadas con valores numericos; por eso el tipo de Python no alcanza para interpretar una variable.

```python
significado_obesidad = {
    'Gender': 'genero', 'Age': 'edad en anos', 'Height': 'altura en metros',
    'Weight': 'peso en kilogramos', 'family_history_with_overweight': 'antecedentes familiares',
    'FAVC': 'consumo frecuente de alimentos caloricos', 'FCVC': 'frecuencia de vegetales',
    'NCP': 'cantidad de comidas principales', 'CAEC': 'consumo entre comidas',
    'SMOKE': 'fuma', 'CH2O': 'consumo de agua', 'SCC': 'control de calorias',
    'FAF': 'frecuencia de actividad fisica', 'TUE': 'uso de dispositivos',
    'CALC': 'consumo de alcohol', 'MTRANS': 'medio de transporte',
    'NObeyesdad': 'nivel de obesidad',
}
tipos_obesidad = pd.DataFrame({
    'tipo_python': obesity.dtypes.astype(str),
    'valores_distintos': obesity.nunique(dropna=False),
    'faltantes': obesity.isna().sum(),
    'significado': pd.Series(significado_obesidad),
})
display(tipos_obesidad)
```

**Interpretacion:** `Gender`, `MTRANS`, `CAEC`, `CALC` y `NObeyesdad` son cualitativas. `Age`, `Height` y `Weight` son cuantitativas. Variables como `FCVC`, `NCP` y `FAF` pueden tener valores decimales porque fueron registradas mediante una escala o generadas sinteticamente, aunque su significado original sea ordinal o discreto.

## 2. Netflix titles

```python
descripcion_netflix = archivos['netflix_txt'].read_text(encoding='utf-8')
print(descripcion_netflix)

print('Al leer un CSV hay que revisar separador, codificacion, comillas y cantidad de columnas.')
netflix_con_columnas_vacias = pd.read_csv(archivos['netflix_csv'], encoding='latin-1')
print('Dimensiones originales:', netflix_con_columnas_vacias.shape)
print('Columnas originales:', list(netflix_con_columnas_vacias.columns))

columnas_vacias = [
    columna for columna in netflix_con_columnas_vacias.columns
    if columna.startswith('Unnamed:') and netflix_con_columnas_vacias[columna].isna().all()
]
netflix = netflix_con_columnas_vacias.drop(columns=columnas_vacias)
print('Columnas vacias eliminadas:', columnas_vacias)
print('Dimensiones limpias:', netflix.shape)
display(netflix.head())
display(netflix.describe(include='all').T)
```

El error UTF-8 se resuelve leyendo con `latin-1`, que es compatible con los caracteres presentes en este archivo. Las columnas `Unnamed:` son columnas extra completamente vacias, por lo que se eliminan despues de verificar que realmente no contienen datos.

```python
tipos_netflix = pd.DataFrame({
    'tipo_python': netflix.dtypes.astype(str),
    'no_nulos': netflix.notna().sum(),
    'faltantes': netflix.isna().sum(),
    'valores_distintos': netflix.nunique(dropna=False),
})
display(tipos_netflix)

print('Tipos de contenido:')
display(netflix['type'].value_counts(dropna=False))
print('La columna duration mezcla minutos de peliculas con temporadas de series; no debe analizarse como una unica cantidad.')
```

### Preguntas sobre Netflix

```python
directores_distintos = netflix['director'].dropna().nunique()
print('Directores distintos:', directores_distintos)

peliculas = netflix[netflix['type'].eq('Movie')].copy()
series = netflix[netflix['type'].eq('TV Show')].copy()
conteo_directores = peliculas['director'].dropna().value_counts()
print('Director con mas peliculas:')
display(conteo_directores.head(1))
print('Cantidad de peliculas:', len(peliculas))
print('Cantidad de series/programas de TV:', len(series))

netflix['release_year'] = pd.to_numeric(netflix['release_year'], errors='coerce')
print('Peliculas mas viejas:')
display(peliculas.loc[peliculas['release_year'].idxmin(), ['title', 'release_year']])
print('Series mas viejas:')
display(series.loc[series['release_year'].idxmin(), ['title', 'release_year']])
```

```python
def cantidad_de_peliculas_por_pais(df, pais):
    paises = (
        df.loc[df['type'].eq('Movie'), 'country']
        .dropna()
        .astype(str)
        .str.split(',')
        .explode()
        .str.strip()
    )
    return paises.eq(pais).sum()

print('Peliculas con participacion de Argentina:', cantidad_de_peliculas_por_pais(netflix, 'Argentina'))
```

Contar con `country == 'Argentina'` solo encuentra filas donde Argentina es el unico texto de la celda. Separar los paises enumerados permite contar tambien las coproducciones. La eleccion depende de la pregunta: para participacion argentina conviene separar y expandir los valores.

## 3. Catalogo de streaming de Argentina 2024

```python
catalogo = pd.read_excel(archivos['catalogo_xlsx'])
print('Dimensiones:', catalogo.shape)
print('Columnas:', list(catalogo.columns))
display(catalogo.head())
catalogo.info()
display(catalogo.describe(include='all').T)
```

El analisis de tipos combina `info()` con el significado de las columnas: un cero puede ser un valor real o un faltante codificado, mientras que un valor nulo expresa ausencia de informacion. La unidad de observacion debe confirmarse revisando si una misma obra aparece en varias plataformas.

```python
print('Valores faltantes por columna:')
display(catalogo.isna().sum().sort_values(ascending=False).to_frame('faltantes'))

if 'platform' in catalogo.columns:
    print('Cantidad de registros por plataforma:')
    display(catalogo['platform'].value_counts())
    print('Cantidad de titulos distintos por plataforma:')
    display(catalogo.groupby('platform')['title'].nunique().sort_values(ascending=False))

if 'genres' in catalogo.columns:
    generos = (
        catalogo['genres'].dropna().astype(str)
        .str.replace(r'[\[\]]', '', regex=True)
        .str.split(r',|;|\|')
        .explode().str.strip()
    )
    print('Generos mas frecuentes:')
    display(generos[generos.ne('')].value_counts().head(10))

columnas_puntaje = [columna for columna in ['imdbScore', 'tmdbScore'] if columna in catalogo.columns]
if 'platform' in catalogo.columns and columnas_puntaje:
    display(catalogo.groupby('platform')[columnas_puntaje].mean().sort_values(columnas_puntaje[0], ascending=False))
```

### Conclusiones del practico

- Antes de analizar hay que conocer la codificacion, las columnas vacias, los tipos y los valores faltantes.
- Las variables categoricas pueden necesitar `value_counts()` o una expansion con `explode()` antes de agrupar.
- Las diferencias entre plataformas describen asociaciones del catalogo; no prueban que una plataforma sea mejor en sentido general.
- La columna `duration` de Netflix debe separarse en minutos para peliculas y cantidad de temporadas para series.

## Preguntas tipo parcial

### 1. Obesidad

**a)** Que `FCVC`, `NCP` y `FAF` aparezcan con decimales puede deberse a una escala ordinal, interpolacion o generacion de datos sinteticos. No es automaticamente un error, pero limita la interpretacion: conviene contrastar los valores observados con la documentacion y no tratarlos como mediciones continuas sin justificacion.

**b)** Primero contaria duplicados exactos y revisaria si son registros repetidos o participantes distintos. Las binarias almacenadas como texto se pueden mapear a `True`/`False`. Las ordinales deben conservar su orden cuando las categorias tienen progresion natural; para variables nominales como `MTRANS` usaria one-hot encoding.

**c)** Tres hipotesis posibles son: edad y peso se relacionan; altura y peso se relacionan; frecuencia de actividad fisica y peso se relacionan. Se evaluan con inspeccion grafica, correlacion de Pearson si la relacion es aproximadamente lineal, y comparacion de grupos cuando una variable sea categorica.

**d)** Las conclusiones solo pueden generalizarse con cuidado a poblaciones parecidas a la muestra. El origen geografico de los participantes introduce sesgo y la presencia mayoritaria de registros sinteticos limita la representacion de personas reales.

### 2. Netflix

**a)** El error UTF-8 es un problema de codificacion del archivo; se verifica probando `latin-1` y revisando caracteres y cantidad de columnas. Las columnas `Unnamed:` completamente vacias suelen surgir por separadores finales o columnas exportadas sin nombre. Se eliminan solo despues de comprobar que son totalmente nulas.

**b)** `duration` mezcla dos unidades: minutos para peliculas y temporadas para programas de TV. La transformacion correcta crea `duration_minutes` solo para peliculas y `duration_seasons` solo para series.

**c)** No eliminaria todos los registros incompletos: perderia informacion y sesgaria el catalogo. La falta de `director` parece depender del tipo de contenido, por lo que conviene informar faltantes por grupo y conservar filas utiles para preguntas que no requieran esa columna.

**d)** La igualdad exacta cuenta solo celdas cuyo unico valor es Argentina. Separar por comas y usar `explode()` cuenta tambien coproducciones; esa estrategia responde mejor a la pregunta sobre participacion argentina.

**e)** Algunos analisis exploratorios son comparar cantidad de peliculas y series por ano, ordenar directores por cantidad de titulos y comparar los paises mas frecuentes separando coproducciones.

### 3. Catalogo argentino

**a)** Si una obra aparece varias veces por estar disponible en distintas plataformas, la unidad de observacion es la disponibilidad obra-plataforma. Contar filas mide disponibilidades; contar titulos distintos mide obras.

**b)** Repetir un titulo puede ser valido si cambia el ano, la obra o la plataforma. La combinacion titulo-plataforma suele ser mas informativa, aunque debe investigarse si se repite. Una clave preferible es un identificador externo de la obra, o una combinacion normalizada de titulo, ano y plataforma cuando no exista ese identificador.

**c)** Para puntajes y votos informaria faltantes y evitaria imputar sin justificar; para `genres` o `director` se puede conservar una categoria `Desconocido`. Un `runtime = 0` no equivale a nulo: hay que revisar documentacion y registros originales para decidir si es un valor valido o un faltante codificado.

**d)** Si la media supera mucho a la mediana y el maximo es muy alto, la distribucion es asimetrica y tiene valores extremos. No eliminaria outliers automaticamente: primero verificaria si son errores o peliculas populares reales, y consideraria transformaciones como `log1p` para votos.

**e)** `year_range` se puede separar con una expresion regular en `year_inicio` y `year_fin`; cuando ambos coinciden se crea `year_puntual`. Asi se distinguen peliculas de un ano de obras con un rango temporal.
