# Practico 3: Acceso a datos y analisis

Se extiende el Practico 2 con profiling automatico, hipotesis bivariadas, correlaciones, scatter plots y rankings de plataformas. Una correlacion describe asociacion lineal, no causalidad.

```python
from pathlib import Path
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from IPython.display import display
from wget import download
from ydata_profiling import ProfileReport

DATA_DIR = Path('datos')
REPORT_DIR = Path('reportes')
DATA_DIR.mkdir(exist_ok=True)
REPORT_DIR.mkdir(exist_ok=True)

def descargar_si_falta(url, nombre):
    destino = DATA_DIR / nombre
    if not destino.exists():
        print(f'Descargando {nombre}...')
        download(url, out=str(destino))
    return destino

obesity_csv = descargar_si_falta(
    'https://ignaciorlando.github.io/datasets/data-science/ObesityDataSet_raw_and_data_sinthetic.csv',
    'obesity.csv'
)
netflix_csv = descargar_si_falta(
    'https://ignaciorlando.github.io/datasets/data-science/netflix_titles.csv',
    'netflix_titles.csv'
)
catalogo_xlsx = descargar_si_falta(
    'https://ignaciorlando.github.io/datasets/data-science/catalogo_streaming_argentina_2024.xlsx',
    'catalogo_streaming_argentina_2024.xlsx'
)

obesity = pd.read_csv(obesity_csv)
netflix = pd.read_csv(netflix_csv, encoding='latin-1')
netflix = netflix.drop(columns=[
    columna for columna in netflix.columns
    if columna.startswith('Unnamed:') and netflix[columna].isna().all()
])
catalogo = pd.read_excel(catalogo_xlsx)
print('Datasets cargados:', obesity.shape, netflix.shape, catalogo.shape)
```

## 1. Perfil y analisis: obesidad

```python
print('Resumen estadistico de obesidad:')
display(obesity.describe(include='all').T)

reporte_obesidad = ProfileReport(
    obesity,
    title='Perfil del conjunto de obesidad',
    explorative=True
)
ruta_reporte_obesidad = REPORT_DIR / 'perfil_obesidad.html'
reporte_obesidad.to_file(output_file=str(ruta_reporte_obesidad))
print(f'Reporte guardado en: {ruta_reporte_obesidad}')
```

El reporte permite contrastar histogramas con `describe()`. Por ejemplo, la media y la mediana ayudan a detectar asimetrias, mientras que los histogramas muestran concentracion, colas y posibles valores extremos. Las variables ordinales pueden mostrar valores decimales por la forma en que fueron registradas o generadas.

```python
def correlacion_y_grafico(df, columna_x, columna_y, hipotesis):
    datos = df[[columna_x, columna_y]].apply(pd.to_numeric, errors='coerce').dropna()
    if len(datos) < 3:
        print(f'No hay datos suficientes para {columna_x} y {columna_y}')
        return
    r, p_value = stats.pearsonr(datos[columna_x], datos[columna_y])
    print(hipotesis)
    print(f'n={len(datos)}, r={r:.4f}, p-value={p_value:.4g}')
    plt.figure(figsize=(6, 4))
    plt.scatter(datos[columna_x], datos[columna_y], alpha=0.35)
    plt.xlabel(columna_x)
    plt.ylabel(columna_y)
    plt.title(f'{columna_x} y {columna_y}: r={r:.3f}')
    plt.grid(True)
    plt.show()
    return r, p_value

hipotesis_obesidad = [
    ('Age', 'Weight', 'Hipotesis: la edad se relaciona con el peso.'),
    ('Height', 'Weight', 'Hipotesis: la altura se relaciona positivamente con el peso.'),
    ('FAF', 'Weight', 'Hipotesis: la actividad fisica se relaciona con el peso.'),
]
for columna_x, columna_y, hipotesis in hipotesis_obesidad:
    correlacion_y_grafico(obesity, columna_x, columna_y, hipotesis)
```

Estas hipotesis son plausibles como preguntas exploratorias, pero no se aceptan o rechazan solo por el signo de `r`. Hay que mirar el scatter plot, el valor p, los extremos y el modo en que se recolectaron los datos. Ademas, correlacion no implica causalidad.

## 2. Perfil y analisis: Netflix

```python
netflix['release_year'] = pd.to_numeric(netflix['release_year'], errors='coerce')
netflix['duration_number'] = pd.to_numeric(
    netflix['duration'].astype(str).str.extract(r'(\d+)')[0],
    errors='coerce'
)
netflix['duration_minutes'] = netflix['duration_number'].where(netflix['type'].eq('Movie'))
netflix['duration_seasons'] = netflix['duration_number'].where(netflix['type'].eq('TV Show'))
netflix['title_length'] = netflix['title'].fillna('').str.len()
netflix['description_length'] = netflix['description'].fillna('').str.len()

display(netflix[['release_year', 'duration_minutes', 'duration_seasons', 'title_length', 'description_length']].describe().T)
reporte_netflix = ProfileReport(
    netflix,
    title='Perfil del catalogo Netflix',
    explorative=True
)
ruta_reporte_netflix = REPORT_DIR / 'perfil_netflix.html'
reporte_netflix.to_file(output_file=str(ruta_reporte_netflix))
print(f'Reporte guardado en: {ruta_reporte_netflix}')
```

```python
peliculas = netflix[netflix['type'].eq('Movie')].copy()

hipotesis_netflix = [
    (peliculas, 'release_year', 'duration_minutes', 'Hipotesis: el ano de estreno se relaciona con la duracion de las peliculas.'),
    (netflix, 'release_year', 'title_length', 'Hipotesis: el ano de estreno se relaciona con la longitud del titulo.'),
    (netflix, 'title_length', 'description_length', 'Hipotesis: titulos mas largos se relacionan con descripciones mas largas.'),
]
for df, columna_x, columna_y, hipotesis in hipotesis_netflix:
    correlacion_y_grafico(df, columna_x, columna_y, hipotesis)
```

La duracion se separo en dos variables porque `90 min` y `2 Seasons` no representan la misma magnitud. Las hipotesis sobre longitud de titulo son exploratorias y no tienen una interpretacion causal directa; sirven para practicar la construccion, medicion y critica de una hipotesis.

## 3. Perfil y analisis: catalogo argentino

```python
print('Resumen estadistico del catalogo:')
display(catalogo.describe(include='all').T)

reporte_catalogo = ProfileReport(
    catalogo,
    title='Perfil del catalogo de streaming argentino 2024',
    explorative=True
)
ruta_reporte_catalogo = REPORT_DIR / 'perfil_catalogo_argentina.html'
reporte_catalogo.to_file(output_file=str(ruta_reporte_catalogo))
print(f'Reporte guardado en: {ruta_reporte_catalogo}')

print('Faltantes por columna:')
display(catalogo.isna().sum().sort_values(ascending=False).to_frame('faltantes'))
```

```python
def correlacion_catalogo(columna_x, columna_y, hipotesis):
    if columna_x not in catalogo.columns or columna_y not in catalogo.columns:
        print(f'Se omite: faltan {columna_x} o {columna_y}')
        return
    correlacion_y_grafico(catalogo, columna_x, columna_y, hipotesis)

hipotesis_catalogo = [
    ('imdbVotes', 'imdbScore', 'Hipotesis: las obras con mas votos tienden a tener un puntaje IMDb diferente.'),
    ('tmdbPopularity', 'tmdbScore', 'Hipotesis: la popularidad TMDB se relaciona con el puntaje TMDB.'),
    ('runtime', 'imdbScore', 'Hipotesis: la duracion se relaciona con el puntaje IMDb.'),
]
for columna_x, columna_y, hipotesis in hipotesis_catalogo:
    correlacion_catalogo(columna_x, columna_y, hipotesis)
```

```python
if 'year_range' in catalogo.columns:
    anos = catalogo['year_range'].astype(str).str.extractall(r'(\d{4})')[0].unstack()
    catalogo['year_inicio'] = pd.to_numeric(anos[0], errors='coerce')
    catalogo['year_fin'] = pd.to_numeric(anos[1], errors='coerce') if 1 in anos.columns else catalogo['year_inicio']
    catalogo['year_puntual'] = catalogo['year_inicio'].where(catalogo['year_inicio'].eq(catalogo['year_fin']))
    display(catalogo[['year_range', 'year_inicio', 'year_fin', 'year_puntual']].head())
    print('La transformacion permite distinguir un ano puntual de un rango.')
```

### Ranking de plataformas

```python
if 'platform' in catalogo.columns:
    ranking = catalogo.groupby('platform').agg(
        registros=('title', 'size'),
        titulos_distintos=('title', 'nunique')
    ).sort_values('titulos_distintos', ascending=False)
    print('Plataformas ordenadas por cantidad de titulos distintos:')
    display(ranking)

    if 'genres' in catalogo.columns:
        generos = catalogo[['platform', 'genres']].dropna().copy()
        generos['genre'] = (
            generos['genres'].astype(str)
            .str.replace(r'[\[\]]', '', regex=True)
            .str.split(r',|;|\|')
        )
        generos = generos.explode('genre')
        generos['genre'] = generos['genre'].str.strip()
        horror = generos[generos['genre'].str.contains('horror', case=False, na=False)]
        print('Plataformas con mas titulos de terror:')
        display(horror.groupby('platform').size().sort_values(ascending=False))
else:
    print('El dataset no tiene una columna platform con ese nombre.')
```

## Conclusiones

- Los histogramas y `describe()` se complementan: uno muestra la forma de la distribucion y el otro resume valores centrales y extremos.
- Los valores extremos no se eliminan automaticamente: primero hay que determinar si son errores, casos validos o parte de la poblacion.
- Los perfiles, correlaciones y rankings son herramientas exploratorias; la interpretacion debe considerar faltantes, unidad de observacion, sesgos y origen de los datos.
- Los reportes HTML quedan en `reportes/` para abrirlos desde el explorador de archivos.
